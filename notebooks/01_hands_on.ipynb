```python

from src.pretty_printer import PrettyPrinter
import gymnasium as gym
```

```python
printer = PrettyPrinter(
    observation_labels=[
        "Position du chariot",
        "Vitesse du chariot",
        "Angle de la barre",
        "Vitesse angulaire",
    ],
    observation_units=[
        "m",
        "m/s",
        "rad",
        "rad/s",
    ],
    observation_descriptions=[
        "Position horizontale du chariot.",
        "Vitesse horizontale du chariot.",
        "Inclinaison de la barre par rapport à la verticale.",
        "Vitesse de rotation de la barre.",
    ],
    action_labels={
        0: "Pousser vers la gauche",
        1: "Pousser vers la droite",
    },
)
```

```python
env = gym.make("CartPole-v1")

printer.env(env)
```

```python
printer.observation_space(env.observation_space)
printer.action_space(env.action_space)
```

```python
observation = env.observation_space.sample()

printer.observation(observation)
```

```python
action = env.action_space.sample()

printer.action(action)
```

```python
env = gym.make("CartPole-v1")

number_of_episodes = 1

for episode in range(1, number_of_episodes + 1):
    observation, info = env.reset()

    total_reward = 0
    reward = 0
    step_number = 0
    terminated = False
    truncated = False

    printer.episode_start(
        episode=episode,
        observation=observation,
    )

    while(not terminated and not truncated):
        step_number += 1


        observation_before = observation.copy()
        action = env.action_space.sample()
        observation, reward, terminated, truncated, info = env.step(action)
        total_reward += reward

        printer.step_compact(
            step_number=step_number,
            action=action,
            reward=reward,
            total_reward=total_reward,
            terminated=terminated,
            truncated=truncated,
        )

env.close()
```
